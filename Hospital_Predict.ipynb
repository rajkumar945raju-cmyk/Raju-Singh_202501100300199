```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('/content/hospital_readmission.csv')
```

```python
df.shape
```

```python
df.info()
```

```python
numeric_col = ['length_of_stay_days','num_diagnoses','num_medications',
               'prev_admissions','glucose_level','bmi','has_diabetes','discharge_type']
```

```python
df[numeric_col] = df[numeric_col].fillna(df[numeric_col].median())
df['age'] = df['age'].fillna(df['age'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
df = pd.get_dummies(df,columns=['age'],drop_first = True)
```

```python
df = df.astype(int)
```

```python
df.head()
```

```python
x = df.drop(columns=['readmitted_30days'],axis=1)
y = df['readmitted_30days']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
col = ['length_of_stay_days','num_diagnoses','num_medications','prev_admissions','glucose_level','bmi']
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
df = df.astype(int)
```

```python
df.head()
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(penalty='l2',C=0.1,max_iter=1000)
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
```

```python
y_prob = model.predict_proba(x_test)[:,1]
```

```python
from sklearn.metrics import roc_auc_score,confusion_matrix,accuracy_score,classification_report
auc = roc_auc_score(y_test,y_prob)
print("ROC-AUC : ",auc)
```

```python
print("Confusion Matrix : ")
print(confusion_matrix(y_test,y_pred))
```

```python
print("Classification Matrix : ")
print(classification_report(y_test,y_pred))
```
